```python
import pandas as pd
import openpyxl
import numpy as np
```

```python
df = pd.read_excel("dataset_merged.xlsx")
df.head()
```

```python
#Добавляем колонку с количеством комнат
df["Количество_комнат"] = df['Название'].str.split(',').str[0]
df['Количество_комнат'] = df['Количество_комнат'].str.replace('-к. квартира', '').str.strip()
df['Количество_комнат']
```

```python
#Добавляем колонку с площадью
df['Площадь'] = df['Название'].str.split(',').str[1]
df['Площадь'] = df['Площадь'].str.replace('м²', '').str.strip()
df['Площадь']
```

```python
# Добовляем колонку с этажом
df['Этаж'] = df['Название'].str.split(',').str[2]
# Извлекаем только число 

df['Этаж'] = df['Этаж'].str.extract(r'(\d+)', expand=False)

# Преобразуем в числовой формат
df['Этаж'] = pd.to_numeric(df['Этаж'], errors='coerce')

df['Этаж']
```

```python
#Добавление этажа и этажности дома 

import re

extracted = df['Название'].str.extract(r'(\d+)\s*/\s*(\d+)\s*эт', expand=True)

extracted.columns = ['Этаж', 'Этажность_дома']

for col in ['Этаж', 'Этажность_дома']:
    extracted[col] = pd.to_numeric(extracted[col], errors='coerce')

df[['Этаж', 'Этажность_дома']] = extracted

df[['Название', 'Этаж', 'Этажность_дома']].head()
```

```python
df_clear['Площадь'] = df_clear['Площадь'].astype(int)
df_clear['Цена'] = df_clear['Цена'].astype(int)
df_clear['Цена_за_квадратный_метр'] = df_clear['Цена'] / df_clear['Площадь']
```

```python
df_clear['Курортный'].replace('N', False, inplace=True)
df_clear['Курортный'].replace('Y', True, inplace=True)
df_clear
```

```python
# Функция для извлечения числа комнат 
def get_rooms_count(text):
    text_lower = str(text).lower() 
    
    
    if 'студия' in text_lower:
        return 0  # Студию считаем за 0 комнат
    if 'своб. планировка' in text_lower:
        return 1  # Свободную планировку условно считаем за 1 (как одно пространство)
    if '10 и более' in text_lower:
        return 10 # Берем 10
    
    match = re.search(r'^(\d+)', text_lower)
    if match:
        return int(match.group(1)) # .group(1) возвращает то, что в скобках
    
    # Если ничего не нашли (например, 'nan' или пустая строка), возвращаем "пропуск"
    return np.nan
```

```python
df_clear['Rooms_Count'] = df_clear['Количество_комнат'].apply(get_rooms_count)
```

```python
# Тип недвижимости
conditions = [
    df_clear['Количество_комнат'].str.contains('студия', case=False, na=False),
    df_clear['Количество_комнат'].str.contains('Своб. планировка', case=False, na=False),
    df_clear['Количество_комнат'].str.contains('апартаменты', case=False, na=False)
]

choices = [
    'Студия',
    'Своб. планировка',
    'Апартаменты'
]

# Создаем столбец: если ни одно условие не сработало, ставим 'Квартира'
df_clear['Property_Type'] = np.select(conditions, choices, default='Квартира')
```

```python
# One-Hot Encoding

df_final = pd.get_dummies(df_clear, 
                          columns=['Property_Type'], 
                          drop_first=True) 


df_final = df_final.drop(columns=['Количество_комнат'])
```

# После EDA

```python
# Обработка даты
df_final['Дата публикации'] = pd.to_datetime(df['Дата публикации'], errors='coerce')
df_final['Year_Public'] = df_final['Дата публикации'].dt.year
df_final['Month_Public'] = df_final['Дата публикации'].dt.month
```

```python
# Преобразование всех булевых колонок
bool_cols = df_final.select_dtypes(include='bool').columns
df_final[bool_cols] = df_final[bool_cols].astype(int)
```

```python
# Логарифмическое преобразование (Борьба с асимметрией) 
# Отчет показал высокую асимметрию для цен и площади
# Используем np.log1p для избежания проблем с нулями

df_final['Цена_за_квадратный_метр'] = np.log1p(df_final['Цена_за_квадратный_метр'])
df_final['Площадь'] = np.log1p(df_final['Площадь'])
```

```python
# Заменяем возможные inf (если были -1 цены, что маловероятно, но все же)
df_final.replace([np.inf, -np.inf], np.nan, inplace=True)
# Удаляем строки, где целевая переменная или дата стали NaN
df_final = df_final.dropna(subset=['Цена_за_квадратный_метр', 'Year_Public'])
```

```python
print(f"Размер после финальной очистки: {df_final.shape}")
```

```python
df_final['Цена'] = df_final['Цена'].astype(int)
```

```python
# Удаляем объекты с ценой выше 40 млн рублей
price_limit = 40000000
df_final = df_final.query(f"Цена <= {price_limit}")
```
